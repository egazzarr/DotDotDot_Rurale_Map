# Count data-point folders by region

Expected structure:

```text
ROOT_FOLDER/
├── Abruzzo/
│   ├── datapoint_001/
│   ├── datapoint_002/
│   └── ...
├── Basilicata/
│   ├── datapoint_001/
│   └── ...
└── ... 20 region folders
```

The notebook counts only the **immediate child folders inside each region folder** as data points.  
This avoids accidentally counting folders nested inside a data-point folder.


````python
from pathlib import Path
import pandas as pd

# Change this to your real outer folder.
ROOT_FOLDER = Path("./0.original_foto_schede/")

# macOS / Linux example:
# ROOT_FOLDER = Path("/Users/yourname/path/to/outside_folder")

ROOT_FOLDER
````

````python
# Basic validation
if not ROOT_FOLDER.exists():
    raise FileNotFoundError(f"Folder does not exist: {ROOT_FOLDER}")

if not ROOT_FOLDER.is_dir():
    raise NotADirectoryError(f"Path is not a folder: {ROOT_FOLDER}")

region_folders = sorted(
    folder
    for folder in ROOT_FOLDER.iterdir()
    if folder.is_dir()
)

print(f"Region folders found: {len(region_folders)}")

if len(region_folders) != 20:
    print("Warning: expected 20 region folders.")
````

````python
# Count immediate data-point folders inside each region folder
rows = []

for region_folder in region_folders:
    data_point_folders = sorted(
        folder
        for folder in region_folder.iterdir()
        if folder.is_dir()
    )

    rows.append({
        "region": region_folder.name,
        "data_points": len(data_point_folders),
        "region_path": str(region_folder),
    })

counts_df = pd.DataFrame(rows).sort_values("region").reset_index(drop=True)

display(counts_df)

total_data_points = int(counts_df["data_points"].sum())

print(f"\nTotal data points: {total_data_points:,}")
````

````python
# Optional: show every detected data-point folder
data_points = []

for region_folder in region_folders:
    for data_point_folder in sorted(region_folder.iterdir()):
        if data_point_folder.is_dir():
            data_points.append({
                "region": region_folder.name,
                "data_point": data_point_folder.name,
                "path": str(data_point_folder),
            })

data_points_df = pd.DataFrame(data_points)

display(data_points_df.head(50))
print(f"Detected data-point folders: {len(data_points_df):,}")
````
